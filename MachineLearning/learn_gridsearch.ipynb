# Grid Search for Decision Tree Classifier

This notebook implements a grid search to find the optimal hyperparameters for a decision tree classifier on the IT jobs dataset.

```python
from sklearn import tree
from sklearn.tree import plot_tree
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split, GridSearchCV
import matplotlib.pyplot as plt
import pandas as pd
import pickle
```

```python
# read the input data
data = pd.read_csv('it_jobs_automl_dataset.csv')

# prepare personal data: remove everything except skills
X = data.drop('got_job', axis=1)
X = X.drop('name', axis=1)

# split off the classification column
y = data['got_job']

# split the table into training and test data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1)
```

```python
# prepare classifier
classifier = tree.DecisionTreeClassifier()
count_columns = data.shape[1]

#classifier = classifier.fit(X_train, y_train)

param_grid = {
    'max_depth': range(1, count_columns),
    'max_features': range(1, count_columns),
    'criterion': ["entropy", "gini"]
}

# fit a decision tree to the training data
grid_search = GridSearchCV(estimator=classifier, param_grid=param_grid)
grid_search.fit(X_train, y_train)
```

```python
# Get the best classifier and make predictions
classifier = grid_search.best_estimator_
```

```python
# test the model
y_pred = classifier.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy}')
```

```python
plt.figure(figsize=(18, 15))
plot_tree(classifier, filled=True, feature_names=X.columns,
          class_names=True)
plt.show()
```

```python

with open('model.pkl', 'wb') as handle:
    pickle.dump(classifier, handle, protocol=pickle.HIGHEST_PROTOCOL)
```
