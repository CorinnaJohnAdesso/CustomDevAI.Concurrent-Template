```python
import pickle

# open file, deserialize model
with open('animals.pkl', 'rb') as handle:
    model = pickle.load(handle)

# open file, deserialize labels
with open('animals-labels.pkl', 'rb') as handle:
    label_encoder = pickle.load(handle)
```

```python
import numpy as np
from PIL import Image
img_width, img_height = 400, 300

# Function to predict the class of a single image
def predict_image(image_path):
    try:
        img = Image.open(image_path).convert('RGB')
        img = img.resize((img_width, img_height))
        img_array = np.array(img)
        img_flattened = img_array.reshape(1, -1)  # Reshape for a single image
        prediction = model.predict(img_flattened)
        predicted_class = label_encoder.inverse_transform(prediction)[0]
        return predicted_class
    except (OSError, FileNotFoundError):
        return "Error: Could not load image"
```

```python
# Predict animal on image
new_image_path = "test-images/test1.JPG"
predicted = predict_image(new_image_path)
print(f"The predicted animal for {new_image_path} is: {predicted}")
```
